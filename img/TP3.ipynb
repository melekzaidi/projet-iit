```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
x=pd.read_csv('synthetic_data.csv')
```

```python
import seaborn as sns
sns.boxplot(data=x,orient='v')
```

```python
x.isnull().sum()
```

```python
plt.hist(x['var1'])
plt.hist(x['var2'], alpha=0.8)
plt.show()
```

```python
plt.scatter(x.iloc[: , 0], x.iloc[: ,1],s = 5)
plt.show()
```

```python
import scipy.cluster.hierarchy as sch
z=sch.linkage(x,method="ward")
z
```

```python
np.set_printoptions(suppress=True)
z
```

```python
den=sch.dendrogram(z)
plt.title("dendrogram")
plt.xlabel("Clusters")
plt.ylabel("Euclidean distance")
plt.show()
```

```python
from sklearn.cluster import AgglomerativeClustering
hc = AgglomerativeClustering(n_clusters=2, affinity='euclidean', linkage='ward')
hc.fit(x)
```

```python
colors=['blue','red','green']
for i in range (0,2):
  plt.scatter(x.iloc[hc.labels_ == i , 0], x.iloc[hc.labels_ == i ,1],
            s = 10, c = colors[i]  , label = 'Cluster {}'.format(i+1))
plt.legend()
plt.show()
```

```python
from sklearn.cluster import KMeans
kmeans = KMeans(n_clusters= 2, init= "k-means++" , random_state = 0)
kmeans.fit(x)
```

```python
colors=['blue','red','green']
for i in range (0,2):
  plt.scatter(x.iloc[hc.labels_ == i , 0], x.iloc[hc.labels_ == i ,1],
            s = 10, c = colors[i]  , label = 'Cluster {}'.format(i+1))
plt.legend()
plt.show()
```
